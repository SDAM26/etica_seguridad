```python
# --- Configuración: ruta del archivo SQLite ---
import os

DB_PATH = os.path.abspath("server.db")  # <-- cambia esto si tu DB está en otra ruta
# Ejemplo:
# DB_PATH = "/home/usuario/proyecto/data/server.db"

print("Usando base de datos:", DB_PATH)
if not os.path.exists(DB_PATH):
    print("⚠️  No se encontró el archivo. Cambia DB_PATH a la ruta correcta.")
```

```python
# --- Leer y mostrar tablas `users` y `vault_entries` ---
import sqlite3, pandas as pd
from IPython.display import display, Markdown

if not os.path.exists(DB_PATH):
    raise FileNotFoundError(f"No existe la base: {DB_PATH}")

conn = sqlite3.connect(DB_PATH)
conn.row_factory = sqlite3.Row

# Tabla USERS
try:
    df_users = pd.read_sql_query(
        "SELECT id, email, pwd_hash, salt_auth, salt_user, created_at FROM users", conn
    )
except Exception as e:
    df_users = pd.DataFrame()
    print("No se pudo leer la tabla 'users':", e)

# Tabla VAULT_ENTRIES
try:
    df_vault = pd.read_sql_query(
        "SELECT id, user_id, title, username, url, note, iv, ciphertext, tag, created_at, updated_at FROM vault_entries",
        conn,
    )
except Exception as e:
    df_vault = pd.DataFrame()
    print("No se pudo leer la tabla 'vault_entries':", e)

conn.close()

display(Markdown("## 🧑‍💻 Usuarios (tabla `users`)"))
if df_users.empty:
    display(Markdown("*<vacío o no existe tabla `users`>*"))
else:
    display(df_users)

display(Markdown("## 🔐 Entradas (tabla `vault_entries`)"))
if df_vault.empty:
    display(Markdown("*<vacío o no existe tabla `vault_entries`>*"))
else:
    display(df_vault)

print(f"\nResumen: {len(df_users)} usuario(s), {len(df_vault)} entrada(s) leídas.")
```
